```python
import numpy as np
import pandas as pd
```

#### Виконати дії з векторами і матрицями:

```python
# Створити квадратну матрицю (n×n), заповнити її за правилом з таблиці варіантів.
n = 4
m = 5
matrix = np.zeros((n,n))
for i in range(1, n+1):
    for j in range(1, n+1):
        matrix[i-1,j-1] = (2*i + 1.75*j)**(1 - 0.5*i)
matrix
```

```python
# Знайти детермінант матриці і обернену до неї матрицю.
det = np.linalg.det(matrix)
inv = np.linalg.inv(matrix)
print("Determinant = ", det)
print("Inverse = ", inv)
```

```python
# Задати два вектори з n елементами, заповнені випадковими числами: один – дійсними числами, другий – цілими числами. Обчислити їх суму і добуток.
v1 = np.random.rand(n)
v2 = np.random.randint(1, 100, size=n)
print("Vector 1:", v1)
print("Vector 2:", v2)
print("Summ:", v1 + v2)
print("Product:", v1 * v2)
```

```python
# Задати дві матриці, заповнені випадковими числами: одна розміру n×m, друга – mn. Знайти їх добуток.
matrix_nm = np.random.randint(1, 100, size=n*m).reshape(n,m)
matrix_mn = np.random.randint(1, 100, size=m*n).reshape(m, n)
print("Matrix mxn:", matrix_mn)
print("Matrix nxm:", matrix_nm)
print("Dot product:", matrix_nm @ matrix_mn)
```

```python
# Знайти добуток матриць і векторів (розбити на дві пари, для яких можливо виконання операції).
v_n = np.random.randint(1, 100, size=n)
v_m = np.random.randint(1, 100, size=m)
print("Vector n:", v_n)
print("Vector m:", v_m)
```

```python
print(f"Dot product Vector n * Matrix nm: {v_n @ matrix_nm}")
print(f"Dot product Matrix nm * Vector m: {matrix_nm @ v_m}")
print(f"Dot product Vector m * Matrix mn: {v_m @ matrix_mn}")
print(f"Dot product Matrix mn * Vector n: {matrix_mn @ v_n}")
```
