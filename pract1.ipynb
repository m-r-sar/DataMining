```python
def scale(matrix, axis=None):
    mins = matrix.min(axis=axis)
    maxs = matrix.max(axis=axis)
    if axis == 1:
        mins = mins.reshape(-1, 1)
        maxs = maxs.reshape(-1, 1)
    matrix = (matrix - mins) / (maxs-mins)
    return matrix
```

```python
mtx = np.random.uniform(0, 15, size=(3, 5))
scale(mtx), scale(mtx, 1), scale(mtx, 0)
```

```python
def evclid(dot1, dot2):
    dist = np.sum((dot1 - dot2)**2)
    dist = np.sqrt(dist)
    return dist
```

```python
dot1 = np.array([0,0])
dot2 = np.array([3,3])
evclid(dot1, dot2)
```

```python
def generate_dot(dim):
    dot = np.random.uniform(100, size=(dim, 1))
    return dot
```

```python
dot1_2 = generate_dot(2)
dot2_2 = generate_dot(2)
evclid(dot1_2, dot2_2)
```

```python

for i in range(1, 11):
    dot1 = generate_dot(i)
    dot2 = generate_dot(i)
    print(i, evclid(dot1, dot2))
```

```python
dim_array = [2,5,10,100,200,500,1000]
dist_array = []
for dim in dim_array:
    points = np.random.rand(100, dim)
    distancies = pdist(points, metric='euclidean')
    min_dist = distancies.min()
    max_dist = distancies.max()
    result = (max_dist - min_dist) / min_dist
    dist_array.append(result)
    print(dim, result)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(dist_array)
```
